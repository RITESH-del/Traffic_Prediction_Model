```python
import pyarrow.parquet as pq
from IPython.display import display

path = "../data/interim/traffic_weather_metr_la.parquet"

pf = pq.ParquetFile(path)
batch = next(pf.iter_batches(batch_size=20))
df_sample = batch.to_pandas()

display(df_sample)
print(df_sample.shape)
print(df_sample.dtypes)
```

```python
from pathlib import Path
import pyarrow.parquet as pq

path = Path("../data/interim/traffic_weather_pems_bay.parquet")

print("Exists:", path.exists())
print("File size (GB):", round(path.stat().st_size / 1e9, 2))

pf = pq.ParquetFile(path)

print("Rows:", pf.metadata.num_rows)
print("Row groups:", pf.metadata.num_row_groups)
print("Columns:", pf.schema.names)
print(pf.schema)
```

```python
import pandas as pd

pf = pq.ParquetFile(path)
batch = next(pf.iter_batches(batch_size=20))
df_sample = batch.to_pandas()

display(df_sample)
print(df_sample.dtypes)
```
